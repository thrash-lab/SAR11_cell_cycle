```python
import numpy as np
from scipy import interpolate

def create_time_weighted_distribution(t, Y):
    """
    Create a time-weighted distribution of rates based on linear interpolation.

    Parameters:
    -----------
    t : array-like
        Time points
    Y : array-like
        Rate values corresponding to time points
    num_samples : int, optional
        Number of samples to generate for the time-weighted distribution

    Returns:
    --------
    Y_new : numpy.ndarray
        Array of time-weighted rate values
    """
    # Ensure t and Y are numpy arrays
    t = np.array(t)
    Y = np.array(Y)

    num_samples = len(t)

    # Check if lengths match
    if len(t) != len(Y):
        raise ValueError("Length of t and Y must be the same")

    # Create a linear interpolation function
    interp_func = interpolate.interp1d(t, Y)

    # Generate evenly spaced time points across the entire range
    # This ensures sampling proportional to time intervals
    t_new = np.linspace(t.min(), t.max(), num_samples)

    # Compute interpolated rate values at these time points
    Y_new = interp_func(t_new)

    return Y_new

#Example
# Your original data
t = [0.0, 0.2847222222222222, 2.600694444444444, 5.7749999999999995, 10.846527777777778,
      14.25763888888889, 15.534722222222221, 18.2, 19.295833333333334, 23.295833333333334,
      28.214583333333334, 31.959722222222222, 35.17638888888889, 39.329166666666666,
      42.625, 45.520833333333336, 45.75138888888889, 47.31875, 48.12777777777777,
      50.488194444444446, 52.94583333333333, 53.018750000000004, 53.21180555555555,
      54.64791666666667, 63.325694444444444, 67.11736111111111, 71.7298611111111,
      84.28402777777778, 97.27916666666665]

Y = [0.36156820629402964, 0.35816894947311617, 0.3928807177270238, 0.32988264041514637,
      0.25880804434904725, 0.2526103483156761, 0.21477372919844012, 0.16569250488251436,
      0.13026900561004556, 0.12190041890270512, 0.20700489684791612, 0.3438131529087934,
      0.3649541350210293, 0.25731285687555144, 0.2204534408333385, 0.47547412217604546,
      0.517583148656245, 0.4840882211485346, 0.3054959316780964, -0.38070394957191467,
      -0.32535833372693823, -0.30687625241923194, -0.27698278809398436, -0.22857222817319656,
      -0.038586647851691955, -0.021114306408334418, -0.004160537123007996, 0.0019335250196143375,
      0.041909973895620664]

# Generate time-weighted distribution
Y_new = create_time_weighted_distribution(t, Y)

print(Y_new)
```

```python
def violin_plot_visualization(data_list, x_loc, ax, ax1, ax2, idr_threshold, cd_threshold, marker, mfc):
  idr_mean_list = []
  cd_mean_list = []
  chi_growth_list = []
  chi_stationary_list = []

  for dl_i in range(len(data_list)):
    target_list = data_list[dl_i]

    cell_density_manual_list = []
    ch_index_list = []
    idr_list = []

    for i in range(len(target_list)):
      cdml = np.array(target_list[i][1])
      cil = np.array(target_list[i][3])
      il = np.array(target_list[i][0])

      day_manual_list = np.array(target_list[i][2])
      day_list = np.array(target_list[i][4])

      cdml_filtered = []
      il_filtered = []
      for dl in day_list:
        cdml_filtered.append(np.mean(cdml[day_manual_list==dl]))
        il_filtered.append(np.mean(il[day_manual_list==dl]))
      #cdml = cdml[[dml in day_list for dml in day_manual_list]]
      #il = il[[dml in day_list for dml in day_manual_list]]

      cdml_redistributed = 10**(create_time_weighted_distribution(day_list, np.log10(cdml_filtered)))
      cil_redistributed = create_time_weighted_distribution(day_list, cil)
      il_redistributed = create_time_weighted_distribution(day_list, il_filtered)

      cell_density_manual_list = np.r_[cell_density_manual_list, cdml_redistributed]
      ch_index_list = np.r_[ch_index_list, cil_redistributed]
      idr_list = np.r_[idr_list, il_redistributed]


      #cell_density_manual_list = np.r_[cell_density_manual_list, cdml_filtered]
      #ch_index_list = np.r_[ch_index_list, cil]
      #idr_list = np.r_[idr_list, il_filtered]

    #idr_sel = idr_list[idr_list > np.percentile(idr_list, 50)]
    #ch_index_growth = ch_index_list[idr_list > np.percentile(idr_list, 50)]

    if idr_threshold != 0 and cd_threshold !=0:
      idr_sel = idr_list[idr_list>idr_threshold*np.max(idr_list)]
      ch_index_growth = ch_index_list[idr_list>idr_threshold*np.max(idr_list)]
      cell_density_sel = cell_density_manual_list[cell_density_manual_list>cd_threshold*np.max(cell_density_manual_list)]
      ch_index_stationary = ch_index_list[cell_density_manual_list>cd_threshold*np.max(cell_density_manual_list)]
    elif idr_threshold != 0:
      idr_sel = idr_list[idr_list>idr_threshold*np.max(idr_list)]
      ch_index_growth = ch_index_list[idr_list>idr_threshold*np.max(idr_list)]
      cell_density_sel = cell_density_manual_list[idr_list<=idr_threshold*np.max(idr_list)]
      ch_index_stationary = ch_index_list[idr_list<=idr_threshold*np.max(idr_list)]
    elif cd_threshold != 0:
      idr_sel = idr_list[cell_density_manual_list<=cd_threshold*np.max(cell_density_manual_list)]
      ch_index_growth = ch_index_list[cell_density_manual_list<=cd_threshold*np.max(cell_density_manual_list)]
      cell_density_sel = cell_density_manual_list[cell_density_manual_list>cd_threshold*np.max(cell_density_manual_list)]
      ch_index_stationary = ch_index_list[cell_density_manual_list>cd_threshold*np.max(cell_density_manual_list)]
    else:
      idr_sel = idr_list
      ch_index_growth = ch_index_list
      cell_density_sel = cell_density_manual_list
      ch_index_stationary = ch_index_list

    if len(idr_sel)<3:
      print('Warning for list #'+str(dl_i))
      print('Too little selected values for making violin plot. Using all the values.')
      idr_sel = idr_list
      ch_index_growth = ch_index_list
      cell_density_sel = cell_density_manual_list
      ch_index_stationary = ch_index_list

    violin_parts = ax.violinplot(idr_sel, positions = [x_loc[dl_i]], showmedians=True, showextrema=False, vert=True)
    for partname in ('bodies', 'cmedians'):
      pc = violin_parts[partname]
      if partname == 'bodies':
        for p in pc:
          p.set_facecolor(cm.tab10(3))
          p.set_alpha(0.2)
      else:
        pc.set_edgecolor(cm.tab10(3))
    if mfc != 'none':
      ax.plot(np.random.randn(len(idr_sel))*0.05+x_loc[dl_i],idr_sel,
              'C3'+marker, alpha=0.5)
    else:
      ax.plot(np.random.randn(len(idr_sel))*0.05+x_loc[dl_i],idr_sel,
              'C3'+marker, alpha=0.5, mfc = mfc)
    idr_mean_list.append(np.mean(idr_sel))

    violin_parts = ax1.violinplot(cell_density_sel, positions = [x_loc[dl_i]], showmedians=True, showextrema=False, vert=True)
    for partname in ('bodies', 'cmedians'):
      pc = violin_parts[partname]
      if partname == 'bodies':
        for p in pc:
          p.set_facecolor(cm.tab10(0))
          p.set_alpha(0.2)
      else:
        pc.set_edgecolor(cm.tab10(0))
    if mfc != 'none':
      ax1.plot(np.random.randn(len(cell_density_sel))*0.05+x_loc[dl_i],cell_density_sel,
              'C0'+marker, alpha=0.5)
    else:
      ax1.plot(np.random.randn(len(cell_density_sel))*0.05+x_loc[dl_i],cell_density_sel,
              'C0'+marker, alpha=0.5, mfc = mfc)

    cd_mean_list.append(np.mean(cell_density_sel))

    violin_parts = ax2.violinplot(ch_index_growth, positions = [x_loc[dl_i]], showmedians=True, showextrema=False, vert=True)
    for partname in ('bodies', 'cmedians'):
      pc = violin_parts[partname]
      if partname == 'bodies':
        for p in pc:
          p.set_facecolor(cm.tab10(3))
          p.set_alpha(0.2)
      else:
        pc.set_edgecolor(cm.tab10(3))
    if mfc != 'none':
      ax2.plot(np.random.randn(len(ch_index_growth))*0.05+x_loc[dl_i],ch_index_growth,
              'C3'+marker, alpha=0.5)
    else:
      ax2.plot(np.random.randn(len(ch_index_growth))*0.05+x_loc[dl_i],ch_index_growth,
              'C3'+marker, alpha=0.5, mfc = mfc)

    chi_growth_list.append(np.mean(ch_index_growth))
    violin_parts = ax2.violinplot(ch_index_stationary, positions = [x_loc[dl_i]], showmedians=True, showextrema=False, vert=True)
    for partname in ('bodies', 'cmedians'):
      pc = violin_parts[partname]
      if partname == 'bodies':
        for p in pc:
          p.set_facecolor(cm.tab10(0))
          p.set_alpha(0.2)
      else:
        pc.set_edgecolor(cm.tab10(0))

    if mfc != 'none':
      ax2.plot(np.random.randn(len(ch_index_stationary))*0.05+x_loc[dl_i], ch_index_stationary,
              'C0'+marker, alpha=0.5)
    else:
      ax2.plot(np.random.randn(len(ch_index_stationary))*0.05+x_loc[dl_i], ch_index_stationary,
              'C0'+marker, alpha=0.5, mfc = mfc)
    chi_stationary_list.append(np.mean(ch_index_stationary))

  return (idr_mean_list, cd_mean_list, chi_growth_list, chi_stationary_list)
```

```python
Pyr_Gly_ratio_list_1 = [[Pyr_1000_Gly_200, Pyr_1000_Gly_500, Pyr_1000_Gly_1000], [0,1,2]]
Pyr_Gly_ratio_list_2 = [[Pyr_500_Gly_100, Pyr_500_Gly_1000], [0,3]]

fig, axs = plt.subplots(nrows = 2, figsize = (6,3))

ax = axs[0]
ax1 = ax.twinx()
ax2 = axs[1]

lines = ['.-', 'o--', ]
markers = ['.', 'o']
labels = ['1mM Pyr.', '0.5mM Pyr.']
mfcs = ['', 'none']
l_flag = 0
for pgrl in [Pyr_Gly_ratio_list_1, Pyr_Gly_ratio_list_2]:
  idr_mean_list, cd_mean_list, chi_growth_list, chi_stationary_list = violin_plot_visualization(pgrl[0], pgrl[1],
                                                                                                ax, ax1, ax2, idr_threshold = 0.5, cd_threshold = 0.65,
                                                                                                marker = markers[l_flag], mfc = mfcs[l_flag])

  if mfcs[l_flag]!='none':
    ax.plot(pgrl[1], idr_mean_list, 'C3'+lines[l_flag], alpha = 0.5)
    ax1.plot(pgrl[1], cd_mean_list, 'C0'+lines[l_flag], alpha = 0.5)
    ax2.plot(pgrl[1], chi_growth_list, 'C3'+lines[l_flag], alpha = 0.5, label = 'Growth\n'+labels[l_flag])
    ax2.plot(pgrl[1], chi_stationary_list, 'C0'+lines[l_flag], alpha = 0.5, label = 'Stationary\n'+labels[l_flag])
  else:
    ax.plot(pgrl[1], idr_mean_list, 'C3'+lines[l_flag], alpha = 0.5, mfc='none')
    ax1.plot(pgrl[1], cd_mean_list, 'C0'+lines[l_flag], alpha = 0.5, mfc='none')
    ax2.plot(pgrl[1], chi_growth_list, 'C3'+lines[l_flag], alpha = 0.5, label = 'Growth\n'+labels[l_flag], mfc='none')
    ax2.plot(pgrl[1], chi_stationary_list, 'C0'+lines[l_flag], alpha = 0.5, label = 'Stationary\n'+labels[l_flag], mfc='none')

  l_flag += 1

for a in axs:
  a.grid('on')
  a.set_xticks([0,1,2,3])
  a.set_xticklabels([])

ax2.set_xticklabels(['5:1',
                     '2:1',
                     '1:1',
                     '1:2'])

ax.set_ylim(0,1.5)
ax1.set_yscale('log', base = 10)
ax1.set_ylim(1e8, 1e10)

ax2.set_ylim(0,6)

ax.set_ylabel('(Doubling/day)', color = 'C3')
ax1.set_ylabel('(cells/mL)', color = 'C0')
ax2.set_ylabel('C-H Index')

# Change the color of the left spine
ax.spines['left'].set_color('C3')
ax1.spines['right'].set_color('C0')

ax.tick_params(axis='y', colors='C3')
ax1.tick_params(axis='y', colors='C0')

ax2.set_ylim(bottom = 0)
ax2.legend(bbox_to_anchor = (1.0,1.05), fontsize = 8)

fig.tight_layout()

fig.savefig('/content/Pyr_Gly_ratio_variation.pdf')
fig.show()
```

```python
fig, axs = plt.subplots(nrows = 2, figsize = (6,3))

ax = axs[0]
ax1 = ax.twinx()
ax2 = axs[1]

Total_list = [Glc_125, Glc_250, Glc_500, Glc_2500]

idr_mean_list = []
cd_mean_list = []
chi_growth_list = []
chi_stationary_list = []

for x_loc in [0,1,2,3]:
  target_list = Total_list[x_loc]

  cell_density_manual_list = []
  ch_index_list = []
  idr_list = []

#  print('#0123-'+str(s))
#  print('[', list(idr_manual), ',\n',
#       list(cell_density_manual_list),',\n',
#       list(day_cell_count_list), ',\n',
#        list(ch_index_list), ',\n',
#        list(day_list),'],')

  for i in range(len(target_list)):
    cdml = np.array(target_list[i][1])
    cil = np.array(target_list[i][3])
    il = np.array(target_list[i][0])

    day_manual_list = np.array(target_list[i][2])
    day_list = np.array(target_list[i][4])

    #cdml = cdml[[dml in day_list for dml in day_manual_list]]
    #il = il[[dml in day_list for dml in day_manual_list]]

    #cell_density_manual_list = np.r_[cell_density_manual_list, cdml]
    #ch_index_list = np.r_[ch_index_list, cil]
    #idr_list = np.r_[idr_list, il]

    cdml_filtered = []
    il_filtered = []
    for dl in day_list:
      cdml_filtered.append(np.mean(cdml[day_manual_list==dl]))
      il_filtered.append(np.mean(il[day_manual_list==dl]))

    cdml_redistributed = 10**(create_time_weighted_distribution(day_list, np.log10(cdml_filtered)))
    cil_redistributed = create_time_weighted_distribution(day_list, cil)
    il_redistributed = create_time_weighted_distribution(day_list, il_filtered)

    cell_density_manual_list = np.r_[cell_density_manual_list, cdml_redistributed]
    ch_index_list = np.r_[ch_index_list, cil_redistributed]
    idr_list = np.r_[idr_list, il_redistributed]


  idr_sel = idr_list[cell_density_manual_list<0.65*np.max(cell_density_manual_list)]
  ch_index_growth = ch_index_list[cell_density_manual_list<0.65*np.max(cell_density_manual_list)]

  cell_density_sel = cell_density_manual_list[cell_density_manual_list>0.65*np.max(cell_density_manual_list)]
  ch_index_stationary = ch_index_list[cell_density_manual_list>0.65*np.max(cell_density_manual_list)]

  violin_parts = ax.violinplot(idr_sel, positions = [x_loc], showmedians=True, showextrema=False, vert=True)
  for partname in ('bodies', 'cmedians'):
    pc = violin_parts[partname]
    if partname == 'bodies':
      for p in pc:
        p.set_facecolor(cm.tab10(3))
        p.set_alpha(0.2)
    else:
      pc.set_edgecolor(cm.tab10(3))

  ax.plot(np.random.randn(len(idr_sel))*0.05+x_loc,idr_sel,
           'C3.', alpha=0.5)
  idr_mean_list.append(np.mean(idr_sel))

  violin_parts = ax1.violinplot(cell_density_sel, positions = [x_loc], showmedians=True, showextrema=False, vert=True)
  for partname in ('bodies', 'cmedians'):
    pc = violin_parts[partname]
    if partname == 'bodies':
      for p in pc:
        p.set_facecolor(cm.tab10(0))
        p.set_alpha(0.2)
    else:
      pc.set_edgecolor(cm.tab10(0))

  ax1.plot(np.random.randn(len(cell_density_sel))*0.05+x_loc,cell_density_sel,
           'C0.', alpha=0.5)

  cd_mean_list.append(np.mean(cell_density_sel))

  violin_parts = ax2.violinplot(ch_index_growth, positions = [x_loc], showmedians=True, showextrema=False, vert=True)
  for partname in ('bodies', 'cmedians'):
    pc = violin_parts[partname]
    if partname == 'bodies':
      for p in pc:
        p.set_facecolor(cm.tab10(3))
        p.set_alpha(0.2)
    else:
      pc.set_edgecolor(cm.tab10(3))

  ax2.plot(np.random.randn(len(ch_index_growth))*0.05+x_loc,ch_index_growth,
           'C3.', alpha=0.5)

  chi_growth_list.append(np.mean(ch_index_growth))
  violin_parts = ax2.violinplot(ch_index_stationary, positions = [x_loc], showmedians=True, showextrema=False, vert=True)
  for partname in ('bodies', 'cmedians'):
    pc = violin_parts[partname]
    if partname == 'bodies':
      for p in pc:
        p.set_facecolor(cm.tab10(0))
        p.set_alpha(0.2)
    else:
      pc.set_edgecolor(cm.tab10(0))

  ax2.plot(np.random.randn(len(ch_index_stationary))*0.05+x_loc,ch_index_stationary,
           'C0.', alpha=0.5)
  chi_stationary_list.append(np.mean(ch_index_stationary))

ax.plot([0,1,2,3], idr_mean_list, 'C3.-', alpha = 0.5)
ax1.plot([0,1,2,3], cd_mean_list, 'C0.-', alpha = 0.5)
ax2.plot([0,1,2,3], chi_growth_list, 'C3.-', alpha = 0.5, label = 'Growth')
ax2.plot([0,1,2,3], chi_stationary_list, 'C0.-', alpha = 0.5, label = 'Stationary')

for a in axs:
  a.grid('on')
  a.set_xticks([0,1,2,3])
  a.set_xticklabels([])

ax2.set_xticklabels(['0.125mM Glc.\n0.05mM Gly.',
                     '0.25mM Glc.\n0.1mM Gly.',
                     '0.5mM Glc.\n0.2mM Gly.',
                     '2.5mM Glc.\n1mM Gly.'])

#ax.set_ylim(-0.2,1.5)
ax1.set_yscale('log', base = 10)
ax1.set_ylim(1e3, 1e9)

#ax2.set_ylim(bottom = 0)
ax2.set_yscale('log', base = 2)
ax2.legend(loc = 2, bbox_to_anchor = (1.0,0.7), fontsize = 8)
ax2.set_yticks([2,4,8,16])
ax2.set_yticklabels([2,4,8,16])

ax.set_ylabel('(Doubling/day)', color = 'C3')
ax1.set_ylabel('Saturation\n(cells/mL)', color = 'C0')
ax2.set_ylabel('C-H Index\n(log scale)')

# Change the color of the left spine
ax.spines['left'].set_color('C3')
ax1.spines['right'].set_color('C0')

ax.tick_params(axis='y', colors='C3')
ax1.tick_params(axis='y', colors='C0')

fig.tight_layout()

fig.savefig('/content/Glc_Gly_2_5_1_ratio.pdf')
fig.show()
```

```python
fig, axs = plt.subplots(nrows = 3, figsize = (3.5,5))

ax = axs[0]
ax1 = axs[1]
ax2 = axs[2]

Total_list = [Glc_125, Glc_250, Glc_500, Glc_2500]

idr_mean_list = []
cd_mean_list = []
chi_growth_list = []
chi_stationary_list = []

for x_loc in [0,1,2,3]:
  target_list = Total_list[x_loc]

  cell_density_manual_list = []
  ch_index_list = []
  idr_list = []

#  print('#0123-'+str(s))
#  print('[', list(idr_manual), ',\n',
#       list(cell_density_manual_list),',\n',
#       list(day_cell_count_list), ',\n',
#        list(ch_index_list), ',\n',
#        list(day_list),'],')

  for i in range(len(target_list)):
    cdml = np.array(target_list[i][1])
    cil = np.array(target_list[i][3])
    il = np.array(target_list[i][0])

    day_manual_list = np.array(target_list[i][2])
    day_list = np.array(target_list[i][4])

    #cdml = cdml[[dml in day_list for dml in day_manual_list]]
    #il = il[[dml in day_list for dml in day_manual_list]]

    #cell_density_manual_list = np.r_[cell_density_manual_list, cdml]
    #ch_index_list = np.r_[ch_index_list, cil]
    #idr_list = np.r_[idr_list, il]

    cdml_filtered = []
    il_filtered = []
    for dl in day_list:
      cdml_filtered.append(np.mean(cdml[day_manual_list==dl]))
      il_filtered.append(np.mean(il[day_manual_list==dl]))

    cdml_redistributed = 10**(create_time_weighted_distribution(day_list, np.log10(cdml_filtered)))
    cil_redistributed = create_time_weighted_distribution(day_list, cil)
    il_redistributed = create_time_weighted_distribution(day_list, il_filtered)

    cell_density_manual_list = np.r_[cell_density_manual_list, cdml_redistributed]
    ch_index_list = np.r_[ch_index_list, cil_redistributed]
    idr_list = np.r_[idr_list, il_redistributed]


  idr_sel = idr_list[cell_density_manual_list<0.65*np.max(cell_density_manual_list)]
  ch_index_growth = ch_index_list[cell_density_manual_list<0.65*np.max(cell_density_manual_list)]

  cell_density_sel = cell_density_manual_list[cell_density_manual_list>0.65*np.max(cell_density_manual_list)]
  ch_index_stationary = ch_index_list[cell_density_manual_list>0.65*np.max(cell_density_manual_list)]

  violin_parts = ax.violinplot(idr_sel, positions = [x_loc], showmedians=True, showextrema=False, vert=True)
  for partname in ('bodies', 'cmedians'):
    pc = violin_parts[partname]
    if partname == 'bodies':
      for p in pc:
        p.set_facecolor(cm.tab10(3))
        p.set_alpha(0.2)
    else:
      pc.set_edgecolor(cm.tab10(3))

  ax.plot(np.random.randn(len(idr_sel))*0.05+x_loc,idr_sel,
           'C3.', alpha=0.5)
  idr_mean_list.append(np.mean(idr_sel))

  violin_parts = ax1.violinplot(cell_density_sel, positions = [x_loc], showmedians=True, showextrema=False, vert=True)
  for partname in ('bodies', 'cmedians'):
    pc = violin_parts[partname]
    if partname == 'bodies':
      for p in pc:
        p.set_facecolor(cm.tab10(0))
        p.set_alpha(0.2)
    else:
      pc.set_edgecolor(cm.tab10(0))

  ax1.plot(np.random.randn(len(cell_density_sel))*0.05+x_loc,cell_density_sel,
           'C0.', alpha=0.5)

  cd_mean_list.append(np.mean(cell_density_sel))

  violin_parts = ax2.violinplot(ch_index_growth, positions = [x_loc], showmedians=True, showextrema=False, vert=True)
  for partname in ('bodies', 'cmedians'):
    pc = violin_parts[partname]
    if partname == 'bodies':
      for p in pc:
        p.set_facecolor(cm.tab10(3))
        p.set_alpha(0.2)
    else:
      pc.set_edgecolor(cm.tab10(3))

  ax2.plot(np.random.randn(len(ch_index_growth))*0.05+x_loc,ch_index_growth,
           'C3.', alpha=0.5)

  chi_growth_list.append(np.mean(ch_index_growth))
  violin_parts = ax2.violinplot(ch_index_stationary, positions = [x_loc], showmedians=True, showextrema=False, vert=True)
  for partname in ('bodies', 'cmedians'):
    pc = violin_parts[partname]
    if partname == 'bodies':
      for p in pc:
        p.set_facecolor(cm.tab10(0))
        p.set_alpha(0.2)
    else:
      pc.set_edgecolor(cm.tab10(0))

  ax2.plot(np.random.randn(len(ch_index_stationary))*0.05+x_loc,ch_index_stationary,
           'C0.', alpha=0.5)
  chi_stationary_list.append(np.mean(ch_index_stationary))

ax.plot([0,1,2,3], idr_mean_list, 'C3.-', alpha = 0.5, label = 'Growth')
ax1.plot([0,1,2,3], cd_mean_list, 'C0.-', alpha = 0.5, label = 'Stationary')
ax2.plot([0,1,2,3], chi_growth_list, 'C3.-', alpha = 0.5, label = 'Growth')
ax2.plot([0,1,2,3], chi_stationary_list, 'C0.-', alpha = 0.5, label = 'Stationary')

for a in axs:
  a.grid('on')
  a.set_xticks([0,1,2,3])
  a.set_xticklabels([])

ax2.set_xticklabels(['0.125mM Glc.\n0.05mM Gly.',
                     '0.25mM Glc.\n0.1mM Gly.',
                     '0.5mM Glc.\n0.2mM Gly.',
                     '2.5mM Glc.\n1mM Gly.'], rotation=45)

#ax.set_ylim(-0.2,1.5)
ax1.set_yscale('log', base = 10)
ax1.set_ylim(1e3, 1e9)

ax.legend(fontsize=8)
ax1.legend(fontsize=8)
#ax2.set_ylim(bottom = 0)
ax2.set_yscale('log', base = 2)
#ax2.legend(loc = 2, bbox_to_anchor = (1.0,0.7), fontsize = 8)
ax2.set_yticks([2,4,8,16])
ax2.set_yticklabels([2,4,8,16])

ax.set_ylabel('(Doubling/day)', color = 'C3')
ax1.set_ylabel('Saturation\n(cells/mL)', color = 'C0')
ax2.set_ylabel('C-H Index\n(log scale)')

for pos in ['left', 'top', 'right', 'bottom']:
  ax.spines[pos].set_color('C3')
  ax1.spines[pos].set_color('C0')

ax.tick_params(axis='y', colors='C3')
ax1.tick_params(axis='y', colors='C0')

fig.tight_layout()

fig.savefig('/content/Glc_Gly_2_5_1_ratio_3s.pdf')
fig.show()
```

```python
fig, axs = plt.subplots(nrows = 2, figsize = (6,3))


ax = axs[0]
ax1 = ax.twinx()
ax2 = axs[1]

Total_list_all = [[Temperature_16C, Temperature_20C, Temperature_25C, Temperature_30C],
                  [Temperature_16C_1, Temperature_20C_1, Temperature_25C_1, Temperature_30C_1]]

for i_tl in range(len(Total_list_all)):
  Total_list = Total_list_all[i_tl]
  idr_median_list = []
  cd_median_list = []
  chi_growth_list = []
  chi_stationary_list = []
  for x_loc in [0,1,2,3]:
    target_list = Total_list[x_loc]

    cell_density_manual_list = []
    ch_index_list = []
    idr_list = []

  #  print('#0123-'+str(s))
  #  print('[', list(idr_manual), ',\n',
  #       list(cell_density_manual_list),',\n',
  #       list(day_cell_count_list), ',\n',
  #        list(ch_index_list), ',\n',
  #        list(day_list),'],')

    for i in range(len(target_list)):
      cdml = np.array(target_list[i][1])
      cil = np.array(target_list[i][3])
      il = np.array(target_list[i][0])

      day_manual_list = np.array(target_list[i][2])
      day_list = np.array(target_list[i][4])

      #cdml = cdml[[dml in day_list for dml in day_manual_list]]
      #il = il[[dml in day_list for dml in day_manual_list]]

      cdml_filtered = []
      il_filtered = []
      for dl in day_list:
        cdml_filtered.append(np.mean(cdml[day_manual_list==dl]))
        il_filtered.append(np.mean(il[day_manual_list==dl]))

      cdml_redistributed = 10**(create_time_weighted_distribution(day_list, np.log10(cdml_filtered)))
      cil_redistributed = create_time_weighted_distribution(day_list, cil)
      il_redistributed = create_time_weighted_distribution(day_list, il_filtered)

      cell_density_manual_list = np.r_[cell_density_manual_list, cdml_redistributed]
      ch_index_list = np.r_[ch_index_list, cil_redistributed]
      idr_list = np.r_[idr_list, il_redistributed]

      #cell_density_manual_list = np.r_[cell_density_manual_list, cdml]
      #ch_index_list = np.r_[ch_index_list, cil]
      #idr_list = np.r_[idr_list, il]

    #idr_sel = idr_list[cell_density_manual_list<=0.65*np.max(cell_density_manual_list)]
    #ch_index_growth = ch_index_list[cell_density_manual_list<=0.65*np.max(cell_density_manual_list)]

    idr_sel = idr_list[idr_list>np.percentile(idr_list, 50)]
    ch_index_growth = ch_index_list[idr_list>np.percentile(idr_list, 50)]

    cell_density_sel = cell_density_manual_list[cell_density_manual_list>0.65*np.max(cell_density_manual_list)]
    ch_index_stationary = ch_index_list[cell_density_manual_list>0.65*np.max(cell_density_manual_list)]

    violin_parts = ax.violinplot(idr_sel, positions = [x_loc], showmedians=True, showextrema=False, vert=True)
    for partname in ('bodies', 'cmedians'):
      pc = violin_parts[partname]
      if partname == 'bodies':
        for p in pc:
          p.set_facecolor(cm.tab10(3))
          p.set_alpha(0.2)
      else:
        pc.set_edgecolor(cm.tab10(3))

    if i_tl==0:
      ax.plot(np.random.randn(len(idr_sel))*0.05+x_loc,idr_sel,
              'C3.', alpha = 0.5)
    else:
      ax.plot(np.random.randn(len(idr_sel))*0.05+x_loc,idr_sel,
              'C3o', mfc = 'none', alpha = 0.5)

    idr_median_list.append(np.mean(idr_sel))

    violin_parts = ax1.violinplot(cell_density_sel, positions = [x_loc], showmedians=True, showextrema=False, vert=True)
    for partname in ('bodies', 'cmedians'):
      pc = violin_parts[partname]
      if partname == 'bodies':
        for p in pc:
          p.set_facecolor(cm.tab10(0))
          p.set_alpha(0.2)
      else:
        pc.set_edgecolor(cm.tab10(0))

    if i_tl==0:
      ax1.plot(np.random.randn(len(cell_density_sel))*0.05+x_loc,cell_density_sel,
              'C0.', alpha = 0.5)
    else:
      ax1.plot(np.random.randn(len(cell_density_sel))*0.05+x_loc,cell_density_sel,
              'C0o', mfc = 'none', alpha = 0.5)

    cd_median_list.append(np.mean(cell_density_sel))

    violin_parts = ax2.violinplot(ch_index_growth, positions = [x_loc], showmedians=True, showextrema=False, vert=True)
    for partname in ('bodies', 'cmedians'):
      pc = violin_parts[partname]
      if partname == 'bodies':
        for p in pc:
          p.set_facecolor(cm.tab10(3))
          p.set_alpha(0.2)
      else:
        pc.set_edgecolor(cm.tab10(3))
    if i_tl==0:
      ax2.plot(np.random.randn(len(ch_index_growth))*0.05+x_loc,ch_index_growth,
              'C3.', alpha = 0.5)
    else:
      ax2.plot(np.random.randn(len(ch_index_growth))*0.05+x_loc,ch_index_growth,
              'C3o', mfc = 'none', alpha = 0.5)

    chi_growth_list.append(np.mean(ch_index_growth))

    violin_parts = ax2.violinplot(ch_index_stationary, positions = [x_loc], showmedians=True, showextrema=False, vert=True)
    for partname in ('bodies', 'cmedians'):
      pc = violin_parts[partname]
      if partname == 'bodies':
        for p in pc:
          p.set_facecolor(cm.tab10(0))
          p.set_alpha(0.2)
      else:
        pc.set_edgecolor(cm.tab10(0))
    if i_tl==0:
        ax2.plot(np.random.randn(len(ch_index_stationary))*0.05+x_loc,ch_index_stationary,
                  'C0.', alpha = 0.5)
    else:
      ax2.plot(np.random.randn(len(ch_index_stationary))*0.05+x_loc,ch_index_stationary,
              'C0o', mfc = 'none', alpha = 0.5)

    chi_stationary_list.append(np.mean(ch_index_stationary))

  ax.set_xticks([0,1,2,3])

  if i_tl == 0:
    ax.plot([0,1,2,3], idr_median_list, 'C3.-', alpha = 0.5)
    ax1.plot([0,1,2,3], cd_median_list, 'C0.-', alpha = 0.5)
    ax2.plot([0,1,2,3], chi_growth_list, 'C3.-', alpha = 0.5, label = 'Growth\n0.2mM Gly.')
    ax2.plot([0,1,2,3], chi_stationary_list, 'C0.-', alpha = 0.5, label = 'Stationary\n0.2mM Gly.')
  else:
    ax.plot([0,1,2,3], idr_median_list, 'C3o--', mfc = 'none', alpha = 0.5)
    ax1.plot([0,1,2,3], cd_median_list, 'C0o--', mfc = 'none', alpha = 0.5)
    ax2.plot([0,1,2,3], chi_growth_list, 'C3o--', mfc = 'none', alpha = 0.5, label = 'Growth\n1mM Gly.')
    ax2.plot([0,1,2,3], chi_stationary_list, 'C0o--', mfc = 'none', alpha = 0.5, label = 'Stationary\n1mM Gly.')

for a in axs:
  a.grid('on')
  a.set_xticks([0,1,2,3])
  a.set_xticklabels([])

ax2.set_xticklabels(['16 °C','20 °C','25 °C', '30 °C'])
ax2.legend(bbox_to_anchor = (1.0,1.05), fontsize = 8)
ax.set_ylim(-0.2,1.5)
ax1.set_yscale('log', base = 10)
ax1.set_ylim(1e3, 1e10)

#ax2.set_ylim(0,6)

ax.set_ylabel('(Doubling/day)', color = 'C3')
ax1.set_ylabel('Saturation\n(cells/mL)', color = 'C0')
ax2.set_ylabel('C-H Index')

# Change the color of the left spine
ax.spines['left'].set_color('C3')
ax1.spines['right'].set_color('C0')

ax.tick_params(axis='y', colors='C3')
ax1.tick_params(axis='y', colors='C0')

fig.tight_layout()

fig.savefig('/content/Temperature_variation_Gly.pdf')
fig.show()
```

```python
file_url = "https://drive.google.com/uc?id=1P517V_ymJY8AOXzvOtIu0vzCgYZg4gZ3"
output_path = "/content/CC_2024_12_02_GT_NH4_Ploidy.json"
gdown.download(file_url, output_path, quiet=False)
ploidy_file_1202 = "/content/CC_2024_12_02_GT_NH4_Ploidy.json"

with open(ploidy_file_1202, 'r') as json_file:
    loaded_data = json.load(json_file)

# Now loaded_data contains your list of dictionaries
loaded_X_dict = loaded_data[0]
loaded_gd_dict = loaded_data[1]
loaded_day_list_dict = loaded_data[2]

sample_keys = list(loaded_X_dict.keys())
fig, axss = plt.subplots(ncols=3, figsize = (4.5, 3))
axs = axss.flatten()

sample_ids_list = [['1', '2'],
              ['9','10'], ['7','8']]
color_code = [0, 100, 200]
titles = ['1mM Gly','3mM Gly', '5mM Gly']

for plot_i in range(len(sample_ids_list)):
  sample_ids = sample_ids_list[plot_i]
  for i in range(len(sample_ids)):
    sample_id = sample_ids[i]

    X_range_list = loaded_X_dict[sample_id]
    gd_list = loaded_gd_dict[sample_id]
    day_list = loaded_day_list_dict[sample_id]

    for j in range(len(day_list)):
      day = day_list[j]
      X_range = np.array(X_range_list[j])
      gd = np.array(gd_list[j])
      range_sel_bool = (X_range>0) & (X_range<8)

      axs[plot_i].fill_between(X_range[range_sel_bool], 5*gd[range_sel_bool]+day, np.zeros(range_sel_bool.sum())+day,
                          color = cm.plasma(color_code[plot_i]),
                          linewidth=0, alpha=0.6)
      axs[plot_i].set_xticks([0,4,8])
  axs[plot_i].grid()
  #axs[plot_i].set_ylim(-2,40)

  if plot_i == 0:
    axs[plot_i].set_ylabel('Days')
  else:
    axs[plot_i].set_yticklabels([])

  axs[plot_i].set_xlabel('Ploidy level')

  axs[plot_i].text(3.3,58,titles[plot_i], color = cm.plasma(color_code[plot_i]))

fig.tight_layout()
fig.show()
```

```python
def get_idr_vs_chi(Total_list_all):
  idr_list_all = []
  chi_list_all = []

  for tl in Total_list_all:
      target_list = tl

      cell_density_manual_list = []
      ch_index_list = []
      idr_list = []

    #  print('#0123-'+str(s))
    #  print('[', list(idr_manual), ',\n',
    #       list(cell_density_manual_list),',\n',
    #       list(day_cell_count_list), ',\n',
    #        list(ch_index_list), ',\n',
    #        list(day_list),'],')

      for i in range(len(target_list)):
        cdml = np.array(target_list[i][1])
        cil = np.array(target_list[i][3])
        il = np.array(target_list[i][0])

        day_manual_list = np.array(target_list[i][2])
        day_list = np.array(target_list[i][4])

        cdml_filtered = []
        il_filtered = []
        for dl in day_list:
          cdml_filtered.append(np.mean(cdml[day_manual_list==dl]))
          il_filtered.append(np.mean(il[day_manual_list==dl]))
        #cdml = cdml[[dml in day_list for dml in day_manual_list]]
        #il = il[[dml in day_list for dml in day_manual_list]]

        cell_density_manual_list = np.r_[cell_density_manual_list, cdml_filtered]
        ch_index_list = np.r_[ch_index_list, cil]
        idr_list = np.r_[idr_list, il_filtered]

        #idr_sel = idr_list[cell_density_manual_list<0.65*np.max(cell_density_manual_list)]
        #ch_index_growth = ch_index_list[cell_density_manual_list<0.65*np.max(cell_density_manual_list)]
        idr_sel = idr_list[idr_list>0.05]
        ch_index_growth = ch_index_list[idr_list>0.05]
        #idr_sel = idr_list
        #ch_index_growth = ch_index_list

      #idr_list_all = np.r_[idr_list_all, idr_list]
      #chi_list_all = np.r_[chi_list_all, ch_index_list]
      idr_list_all = np.r_[idr_list_all, idr_sel]
      chi_list_all = np.r_[chi_list_all, ch_index_growth]

  return idr_list_all, chi_list_all
```
